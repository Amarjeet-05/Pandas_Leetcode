```python
import pandas as pd
```

```python
data = [[1, 'Abbot'], [2, 'Doris'], [3, 'Emerson'], [4, 'Green'], [5, 'Jeames'], [6, 'Aman']]
seat = pd.DataFrame(data, columns=['id', 'student']).astype({'id':'Int64', 'student':'object'})
```

```python
seat
```

```python
a = list(seat['student'])
```

```python
a
```

```python
for i in range(0,len(a),2):
    if i == len(a)-1: 
        break
    tem = a[i] 
    a[i] = a[i+1]
    a[i+1] = tem 
    
```

```python
a
```

```python
seat['student'] = a
```
